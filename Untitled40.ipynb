```python
dummy=pd.get_dummies(a[['date','location','river','basin']],drop_first=True)
print(dummy)
```

```python
c=pd.concat([a,dummy],axis=1)
c
```

```python
d=c.drop(['date','location'],axis=1)
d
```

```python
print(d.dtypes)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
```

```python
d['target_next_discharge']=d['river_discharge_m3s'].shift(-1)
d=d.dropna(subset=['target_next_discharge'])
```

```python
x=d.select_dtypes(include=['number','bool']).drop(columns=['river_discharge_m3s','target_next_discharge'],errors='ignore').values
y=d['target_next_discharge'].values
```

```python
split_index=int(len(x)*0.8)
x_train,x_test=x[:split_index],x[split_index:]
y_train,y_test=y[:split_index],y[split_index:]
```

```python
model=RandomForestRegressor(n_estimators=50,random_state=42)
model.fit(x_train,y_train)
pred=model.predict(x_test)
```

```python
print('MAE:',mean_absolute_error(y_test,pred),2)
print('RMSE:',np.sqrt(mean_squared_error(y_test,pred)),2)
print('R2:',r2_score(y_test,pred),4)
```
